```python
import malariagen_data
import os
import plotly.io as pio
pio.renderers.default = "notebook+colab"
```

```python
ag3 = malariagen_data.Ag3()
ag3
```

```python
#This sets the variable of the genomic region of TREH and the no of snps estimated to be within that region
region = "3L: 36,938,555-36,949,962"
n_snps = 1000
```

```python
pca_bf_df, evr_bf = ag3.pca(
    region=region,
    n_snps=n_snps,
    sample_sets="3.0",
);
```

```python
evr_bf
```

```python
#Allows us to see the variance that exist in a bar chart bewteen PCs which will further inform our decision on which PCs to use for the plot
ag3.plot_pca_variance(evr_bf)
```

```python
pca_bf_df.to_excel("uftreh_pca.xlsx")
```

```python
#To filter out samples with taxon labelled "unassigned"
fpca_bf_df= pca_bf_df.query("taxon != 'unassigned'")
fpca_bf_df.to_excel("treh_pca.xlsx")
fpca_bf_df
```

```python
#To plot a two dimensional PCA
ag3.plot_pca_coords(
    fpca_bf_df,
    color="taxon",
    title="Taxonomic structure of TREH"
)
```
