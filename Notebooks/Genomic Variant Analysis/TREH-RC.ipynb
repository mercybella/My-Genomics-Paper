Manually edit the treh_rc_df file to form the treh-rc.xlsx file

```python
import pandas as pd

df = pd.read_excel("treh-rc.xlsx")

# First column = SNP labels
df = df.set_index(df.columns[0])

species_order = ["gamb", "colu", "arab", "pwan", "biss"]

def extract_species(col):
    for sp in species_order:
        if f"_{sp}_" in col:
            return sp
    return "other"

species = [extract_species(c) for c in df.columns]

cols_sorted = [x for _, x in sorted(
    zip(species, df.columns),
    key=lambda x: species_order.index(extract_species(x[1]))
)]

df = df[cols_sorted]
species_sorted = [extract_species(c) for c in df.columns]
import re

df.columns = [
    re.sub(r"_(gamb|colu|arab|pwan|biss)_", "_", col)
    for col in df.columns
]
df = df.apply(pd.to_numeric, errors='coerce')

Pdf = df * 100
RPdf = Pdf.round(0).astype(int)

```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(18,10))

sns.heatmap(RPdf, annot=True, fmt="d", annot_kws={"size": 8}, cmap="Reds", ax=ax, cbar=True)

#ax.set_yticks(range(len(RPdf.index))) 
ax.set_yticks(np.arange(len(RPdf)) + 0.4)
ax.set_yticklabels(RPdf.index, fontsize=9, va="center")
# Bottom labels = FULL column names (unchanged)
ax.set_xticklabels(RPdf.columns, rotation=90, fontsize=7)

# Create top axis
ax_top = ax.twiny()
ax_top.set_xlim(ax.get_xlim())

positions = []
labels = []

start = 0
for sp in species_order:
    count = species_sorted.count(sp)
    if count > 0:
        mid = start + count / 2
        positions.append(mid)
        labels.append(sp)
        start += count

ax_top.set_xticks(positions)
ax_top.set_xticklabels(labels, fontsize=12)

# Move top axis up
ax_top.xaxis.set_ticks_position('top')
ax_top.spines['top'].set_position(('outward', 40))

current = 0
for sp in species_order:
    count = species_sorted.count(sp)
    if count > 0:
        current += count
        ax.axvline(current, color='black', linewidth=2)
             
plt.title("Amino Acid Change Frequencies of TREH-RC (%)")
plt.tight_layout()
plt.savefig("treh-rc.svg", format="svg", bbox_inches="tight")
plt.show()
```
